## f(x) = x^3 - 3x + 1 tem uma única raiz em (1, 2). Calcule-a com e1 = e2 = 10^-1

```python
import numpy as np
import pandas as pd

def funcao(x):
    return x**3 - 3*x + 1

def falsaPosicao(a, b, epsilon):
    historico = []
    iteracoes = 1

    f_a, f_b = funcao(a), funcao(b)
    comprimento = b - a

    while True:
        x_o = (a * f_b - b * f_a) / (f_b - f_a)
        f_xo = funcao(x_o)
        modulo_fxo = abs(round(f_xo, 5))
        comprimento = b - a

        historico.append ({
            "iteração": iteracoes, "a": a, "b": b, "xo": x_o, "f(a)": f_a, "f_b": f_b,
            "f(xo)": f_xo, "b - a": comprimento, "|f(xo)|": modulo_fxo
        })

        if (modulo_fxo <= epsilon or comprimento <= epsilon):
            break

        if (f_a * f_xo < 0):
            b = x_o
            f_b = f_xo
        else:
            a = x_o
            f_a = f_xo

        iteracoes += 1
    df = pd.DataFrame(historico)
    print(df)

falsaPosicao(1, 2, 1e-1)
```
