**FUNÇÃO PARA PLOT**

```python
import matplotlib.pyplot as plt

def plot(funcao, inicio, fim, nome):
    x = np.linspace(inicio,fim, 1000)
    y = funcao(x)

    plt.plot(x, y, color="pink", linewidth=2, label=nome)
    plt.axhline(0, color="black", linewidth=1)
    plt.axvline(0, color="black", linewidth=1)
    plt.grid(True, alpha=0.3)
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title(nome)
    plt.legend()

    plt.show()
    
```

##### **a)** f(x) = *ln(x)*

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.log(x)

a = 0.5
b = 2 

fa = f(a)
fb = f(b)

# encontrar um intervalo onde o produto é menor que zero
print(fa*fb)

i = 0
epsilon = 10**-5
xm = 0.0
fxm = 0.0

while i < 50:
    # bisseção
    xm = (a + b) / 2
    fxm = f(xm)

    if (abs(fxm) < epsilon):
        break

    if fa * fxm < 0:
        b = xm
        fb = fxm
    else:
        a = xm
        fa = fxm

    i += 1

print("Raiz aproximada: ", xm)
print("f(xm): ", fxm)

plot(f, 0.5, 2, "f(x) = ln(x)")
```
