```python
import numpy as np
import sympy as sy
import matplotlib.pyplot as plt

def f(x):
    return x * np.exp(x) - 10

def f_x(x):
    return x * np.exp(x) + np.exp(x)

def fx__(x):
    return x * np.exp(x) + 2 * np.exp(x)

def newtonRapshon(a, b, TOL, n):
    x = (a + b) / 2
    x_ant = float('nan')

    for k in range(1, n + 1):
        epsilon = abs((x - x_ant) / (max(x, 1)))
        print(k, x, f(x), epsilon)

        if (f(x) == 0 or epsilon < TOL):
            print(f"RAIZ ENCONTRADA: {x}")
            break

        x_ant = x
        x = x - (f(x) / f_x(x))
    
    return x

newtonRapshon(1, 2, 1e-5, 100)

def plot(a, b, n):
    intervalo = np.linspace(a, b, n)
    plt.plot(intervalo, f(intervalo), label = 'f(x)', color = 'pink')
    plt.plot(intervalo, f_x(intervalo), label = 'f_(x)', color = 'purple')
    plt.plot(intervalo, fx__(intervalo), label = 'f2(x)', color = 'green')
    plt.axvline(0, linewidth = 1,linestyle = '-', color = 'black')
    plt.axhline(0, linewidth = 1,linestyle = '-', color = 'black')
    plt.grid(True)
    plt.show()

plot(1, 2, 100)
```

```python
# derivada
x_sy = sy.Symbol('x')
funcao = x_sy * sy.exp(x_sy) - 10
df = sy.diff(funcao, x_sy)

print(df)

# segunda derivada
funcao_df = x_sy * sy.exp(x_sy) + sy.exp(x_sy)
df_ = sy.diff(funcao_df, x_sy)

print(df_)
```
