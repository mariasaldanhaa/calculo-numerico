f(x) = 2 * x** 5 − 20 * x**4 + 5*x **3+ x − 1

## Gráfico da função 

```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return 2*x**5 - 20*x**4 + 5*x**3 + x - 1

def plot(a, b):
    intervalos = np.linspace(a, b)
    plt.plot(intervalos, f(intervalos), label = 'f(x)', color = 'purple')
    plt.axhline(0, color = 'black', linestyle = '-', linewidth = 1)
    plt.axvline(0, color = 'black', linestyle = '-', linewidth = 1)
    plt.xlim(a, b)
    plt.title('2 * x** 5 − 20 * x**4 + 5*x **3+ x − 1')
    plt.legend()
    plt.grid()
    plt.show()

plot(-5, 5)
plot(-10, 30)
plot(-2, 15)
```

a função tem raízes nos intervalos [-2, 4] e [8, 10]

## Utilizando o método da Posição Falsa na função f(x)

```python
import pandas as pd

def f_(x):
    return 2*x**5 - 20*x**4 + 5*x**3 + x - 1

def posicaoFalsa(a, b, TOL, n):
    historico = []
    x_ = float('nan')

    for k in range(1, n + 1):
        x = (a * f(b) - b * f(a)) / (f(b) - f(a))
        erro = abs((x - x_) / (max(x, 1)))
        x_ = x

        historico.append({
            "k": k, "a": a, "b": b, "erro": erro,"x": x, "f(x)": f(x), "f(x) x f(a)": f(x)*f(a)
        })

        if (f(x) == 0 or erro < TOL):
            break

        if (f(x) * f(a) > 0):
            a = x
        else:
            b = x
    df = pd.DataFrame(historico)
    print(df)

posicaoFalsa(-2, 15, 1e-6, 100)
```

O erro 0.003082 é maior que 1e-6. Logo, a função diverge na raiz -1.378282.
