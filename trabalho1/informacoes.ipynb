## Velocidade segura em ferrovias: aplicação de métodos numéricos à frenagem de trens pesados

Alunos: Gabriel Lucas Almeida Moreira e Maria Eduarda Saldanha Alves

### Implementando a função nos métodos escolhidos

#### **Cenário base:** frenagem de um trem de minério

Tolerância: 10e-6

Equação do problema:
 
```
f(v) = v·t_r + (m / (2c)) · ln(1 + c·v² / F) − D = 0
```
 
Com `F = m·a` e `c/m` como arrasto específico, a equação também pode ser reescrita como:
 
```
f(v) = v·t_r + (1 / (2·(c/m))) · ln(1 + (c/m)·v² / a) − D = 0
```
 
## Parâmetros
 
| Variável | Valor | Origem |
|---|---|---|
| Massa `m` | 40.000 t (4×10⁷ kg) | Fonte: material da Engineers Australia (trens de minério e carvão acima de 40.000 t) |
| Velocidade máxima de operação | 80 km/h (22,2 m/s) | Fonte: ANTT |
| Desaceleração `a` | 0,3 m/s² (faixa de 0,15 a 0,3) | Fonte didática: exemplo de trem de carga a 80 km/h (Engineers Australia) |
| Tempo de resposta `t_r` | 25 s (faixa de 10 a 35 s) | Escolhida pelos autores |
| Arrasto específico `c/m` | 1,3×10⁻⁵ m⁻¹ | Derivado da equação de Davis modificada (K = 0,07, vagão de ~131 t) |
| Distância livre `D` | 1.000 m (variar até 2.000 m) | Escolhida pelos autores |
 
## Cenários para comparar
 
| Cenário | `a` (m/s²) | `t_r` (s) | Ideia |
|---|---|---|---|
| Base (freio pneumático) | 0,3 | 25 | Referência |
| Freio fraco ou trem carregado | 0,15 | 25 | Sensibilidade a `a` |
| Freio eletrônico (ECP) | 0,3 | 5 | Praticamente elimina o atraso de propagação do sinal |
