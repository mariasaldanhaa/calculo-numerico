### **Exercício 1:** Plote o gráfico e encontre um intervalo contendo uma única raiz das seguintes funções:

#### **a)** f(x) = ln(x)

```python
import numpy as np
import matplotlib.pyplot as plot

x = np.linspace(0.5, 2, 100)
y = np.log(x)

plot.plot(x, y, label="f(x) = ln(x)")

plot.axhline(0, color="black", linewidth=0.8)
plot.axvline(0, color="black", linewidth=0.8)

plot.scatter(1, 0, color="red", zorder=5, label="Raiz: x = 1")

plot.xlabel("x")
plot.ylabel("f(x)")
plot.title("Gráfico de f(x) = ln(x)")

plot.grid(True, linestyle="--", alpha=0.5)
plot.legend()

plot.show()
```

#### **b)** f(x) = eˣ

```python
import numpy as np
import matplotlib.pyplot as plot

x = np.linspace(-2.5, 1, 100)
y = np.exp(x)

plot.plot(x, y, label="f(x) = eˣ")

plot.axhline(0, color="black", linewidth=0.8)
plot.axvline(0, color="black", linewidth=0.8)

# não tem raiz,já que não tem nenhum ponto onde que quando aplicado o x resulte em 0

plot.xlabel("x")
plot.ylabel("f(x)")
plot.title("Gráfico de f(x) = eˣ")

plot.grid(True, linestyle="--", alpha=0.5)
plot.legend()

plot.show()
```

#### **c)** f(x) = cos(x)

```python
import numpy as np
import matplotlib.pyplot as plot

x = np.linspace(-2, 2, 100)
y = np.cos(x)

plot.plot(x, y, label="f(x) = cos(x)")

plot.axhline(0, color="black", linewidth=0.8)
plot.axvline(0, color="black", linewidth=0.8)

plot.scatter(np.pi/2, 0, color="red", zorder=5, label="Raiz: x = π/2")

plot.xlabel("x")
plot.ylabel("f(x)")
plot.title("Gráfico de f(x) = cos(x)")

plot.grid(True, linestyle="--", alpha=0.5)
plot.legend()

plot.show()
```

#### **d)** f(x) = x² + 2x + 1

```python
import numpy as np
import matplotlib.pyplot as plot

x = np.linspace(-2, 2, 100)
y = np.power(x, 2) + 2 * x + 1

plot.plot(x, y, label="f(x) = x² + 2x + 1")

plot.axhline(0, color="black", linewidth=0.8)
plot.axvline(0, color="black", linewidth=0.8)

plot.scatter(-1, 0, color="red", zorder=5, label="Raiz: -1")

plot.xlabel("x")
plot.ylabel("f(x)")
plot.title("Gráfico de f(x) = x² + 2x + 1")

plot.grid(True, linestyle="--", alpha=0.5)
plot.legend()

plot.show()
```

#### **e)** f(x) = e^(2-x²)(x+1)²

```python
import numpy as np
import matplotlib.pyplot as plot

x = np.linspace(-1.5, 3, 100)
y = np.exp(2 - x**2) * (x + 1)**2

plot.plot(x, y, label="f(x) = e^(2-x²)(x+1)²")

plot.axhline(0, color="black", linewidth=0.8)
plot.axvline(0, color="black", linewidth=0.8)

plot.scatter(-1, 0, color="red", zorder=5, label="Raiz: -1")

plot.xlabel("x")
plot.ylabel("f(x)")
plot.title("Gráfico de f(x) = e^(2-x²)(x+1)²")

plot.grid(True, linestyle="--", alpha=0.5)
plot.legend()

plot.show()
```

#### **i)** f(x) = √x - cos(x)

```python
import numpy as np
import matplotlib.pyplot as plot

x = np.linspace(0, 1, 100)
y = np.sqrt(x) - np.cos(x)

plot.plot(x, y, label="f(x) = √x - cos(x)")

plot.axhline(0, color="black", linewidth=0.8)
plot.axvline(0, color="black", linewidth=0.8)

plot.scatter(0.6417140993947, 0, color="red", zorder=5, label="Raiz: 0.6417140993947")

plot.xlabel("x")
plot.ylabel("f(x)")
plot.title("Gráfico de f(x) = √x - cos(x)")

plot.grid(True, linestyle="--", alpha=0.5)
plot.legend()

plot.show()
```
