**FUNÇÃO PARA PLOT**

```python
import numpy as np
import matplotlib.pyplot as plt

def plot(funcao, inicio, fim, nome):
    x = np.linspace(inicio,fim, 1000)
    y = funcao(x)

    plt.plot(x, y, color="pink", linewidth=2, label=nome)
    plt.axhline(0, color="black", linewidth=1)
    plt.axvline(0, color="black", linewidth=1)
    plt.grid(True, alpha=0.3)
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.title(nome)
    plt.legend()

    plt.show()
    
```

#### **Exercício 1:** Utilize o método da posição falsa para calcular a raiz aproximada das seguintes equações. Considere os seguintes pontos:

- Precisão de 6 casas decimais;
- Número máximo de 50 iterações e € < 10-5;
- Isole o intervalo da raiz;
- Analise a convergência do método;
- Apresente o código, solucão e passo a passo para chegar na respostas.


##### **a)** f(x) = ln(x² + 1) - x

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.log(x**2 + 1) - x

# determinando um intervalo
a = -1
b = 1

fa = f(a)
fb = f(b)

# encontrar um intervalo onde o produto é menor que zero
print(fa*fb)

i = 0
epsilon = 10**-5
xm = 0.0
fxm = 0.0

while i < 50:
    # posição false
    xm = (a*fb - b*fa) / (fb - fa)
    fxm = f(xm)

    if (abs(fxm) < epsilon):
        break

    if fa * fxm < 0:
        b = xm
        fb = fxm
    else:
        a = xm
        fa = fxm

    i += 1

print("Raiz aproximada: ", xm)
print("f(xm): ", fxm)

plot(f, -1, 1, "f(x) = ln(x² + 1) - x")
```
