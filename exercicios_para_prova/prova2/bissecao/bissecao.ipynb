# Exemplo 1: 
## f(x) = x^3 - 3x + 1 tem uma unica raiz em (0, 1). Calcule-a com precisao e = 10^-1

```python
import bissecao_cn
import numpy as np

f_x = lambda x: np.power(x, 3) - np.multiply(3, x) + 1
print(bissecao_cn.bissecao(a = 0, b = 1, epsilon = 1e-1, f = f_x))

print(bissecao_cn.grafico(a = 2, b = 3, f =f_x))
```

Raiz aproximada: 0.3125

# Exemplo 2: 
## f(x) = x ln x - 2 tem uma unica raiz em (2, 3). Calcule-a com precisao e = 10^-1

# Exemplo 3:
## Encontre a raiz real da função f(x) = x³ - 2x² - 5x + 6 no intervalo [1.5, 4.0] com precisão ε = 0.1.

```python
import bissecao_cn
import numpy as np

f_x = lambda x: np.power(x, 3) - np.multiply(2, np.power(x, 2)) - np.multiply(5, x) + 6
print(bissecao_cn.bissecao(a = 1.5, b = 4, epsilon = 1e-1, f = f_x))

bissecao_cn.grafico(a = 1.5, b = 4, f =f_x)
```
