```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return x * np.exp(x) - 10

def secante(a, b, TOL, n):
    x_ant = a
    x_pos = b

    for k in range(1, n + 1):
        x = x_pos - f(x_pos) * ((x_pos - x_ant) / (f(x_pos) - f(x_ant)))
        epsilon = abs((x - x_ant) / (max(x, 1)))
        print(k, x, f(x), epsilon)

        if (f(x) == 0 or epsilon < TOL):
            print(f"RAIZ ENCONTRADA: {x}")
            break

        x_ant = x_pos
        x_pos = x

secante(1, 2, 1e-5, 100)

def plot(a, b, n):
    intervalo = np.linspace(a, b, n)
    plt.plot(intervalo, f(intervalo), label = 'f(x)', color = 'pink')
    plt.axvline(0, linestyle = '-', linewidth = 1, label = 'y')
    plt.axhline(0, linestyle = '-', linewidth = 1, label = 'x')
    plt.grid(True)
    plt.show()

plot(1, 2, 100)
```
