f(x) = x*sin(x^2 + 1)

## Gráfico da função 

```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
    return x * np.sin(x ** 2 + 1)

def plot(a, b):
    intervalos = np.linspace(a, b)
    plt.plot(intervalos, f(intervalos), label = 'f(x)', color = 'purple')
    plt.axhline(0, color = 'black', linestyle = '-', linewidth = 1)
    plt.axvline(0, color = 'black', linestyle = '-', linewidth = 1)
    plt.xlim(a, b)
    plt.title('x*sin(x^2 + 1)')
    plt.legend()
    plt.grid()
    plt.show()

plot(0, 1)
plot(0, 3)
plot(0, 5)
```

a função não tem apenas uma única raiz, há vários pontos em que intercepta o eixo x

## Utilizando o método de Newton Raphson na função f(x)

```python
import sympy as sy
import numpy as np
import pandas as pd

x_sy = sy.Symbol('x')
funcao_sy = x_sy * sy.sin(x_sy**2 + 1)
derivada = sy.diff(funcao_sy, x_sy)
print(f"Derivada de f(x): {derivada}")

def f_(x):
    return 2*x**2*np.cos(x**2 + 1) + np.sin(x**2 + 1)

def newtonRaphson(a, b, TOL, n):
    historico = []
    x = (a + b) / 2
    x_ = float('nan')

    for k in range(1, n + 1):
        erro = abs((x - x_) / (max(x, 1)))

        historico.append({
            "k": k, "x": x, "f(x)": f(x), "erro": erro,
        })

        if (f(x) == 0 or erro < TOL):
            break

        x_ = x
        x = x - (f(x) / f_(x))

    df = pd.DataFrame(historico)
    print(df)

newtonRaphson(0, 5, 1e-6, 100)
```
