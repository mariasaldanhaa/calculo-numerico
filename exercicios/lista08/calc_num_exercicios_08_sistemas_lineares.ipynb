```python
import numpy as np
import pandas as pd
import sympy as sy
import resolvedor
```

a) $
\begin{cases}
    2y + 2z = 8 \\
x + 2y +  z = 9 \\
x +  y +  z = 6
\end{cases}
$

```python
# Matriz estendida
mat = [
    [0, 2, 2, 8],
    [1, 2, 1, 9],
    [1, 1, 1, 6]
]
mat = np.array(mat)
print(mat)
resolvedor.resolve(mat)
```

b) $\begin{cases}
x + y + z = 0 \\
x +   10z = -48 \\
10y +  z = 25
\end{cases}$

```python
# matriz estendida
mat = [
    [1, 1, 1, 0],
    [1, 0, 10, -48],
    [0, 10, 1, 25]
]
mat = np.array(mat)
resolvedor.resolve(mat)
```

c) $\begin{cases}
